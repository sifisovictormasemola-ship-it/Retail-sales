```sql
select * from `retail`.`default`.`month_number_from_sale_date` limit 100;
SELECT
    `Month number from sale date` AS sale_id,
    _c1 AS product_name,
    _c2 AS sale_date,
    _c3 AS amount,
    MONTH(TO_DATE(_c2, 'M/d/yy')) AS month_number
FROM `retail`.`default`.`month_number_from_sale_date`
WHERE `Month number from sale date` != 'Sale_id';




SELECT
    orders AS order_id,
    _c1 AS customer_id,
    _c2 AS order_date,
    DAYNAME(TO_DATE(_c2, 'M/d/yy')) AS day_name
FROM `retail`.`default`.`orders`
WHERE orders != 'order_id';
```
